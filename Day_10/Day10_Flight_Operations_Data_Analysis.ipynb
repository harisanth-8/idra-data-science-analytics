## 5. Filtering Records

Filtering is used to answer practical questions about specific flight operations.

```python
# Records where Passengers is greater than or equal to its median
filtered_data = df[df['Passengers'] >= df['Passengers'].median()]
display(filtered_data.head(10))
print('Filtered records:', len(filtered_data))
```

```python
# Filter records belonging to the most common Flight_ID
filtered_category = df[df['Flight_ID'] == 'FL0001']
display(filtered_category.head(10))
print('Records in selected category:', len(filtered_category))
```

## 6. Sorting Records

Sort the dataset by numerical and categorical fields to identify high/low values and common categories.

```python
# Sort by Passengers in descending order
sorted_desc = df.sort_values(by='Passengers', ascending=False)
display(sorted_desc.head(10))
```

```python
# Sort by Passengers in ascending order
sorted_asc = df.sort_values(by='Passengers', ascending=True)
display(sorted_asc.head(10))
```

```python
# Sort using two columns
df.sort_values(by=['Flight_ID', 'Passengers'], ascending=[True, False]).head(10)
```

## 7. Grouping and Aggregation

Group the data by categorical features and calculate counts, averages, totals, minimums, and maximums where appropriate.

```python
# Count records by Flight_ID
group_count = df.groupby('Flight_ID').size().sort_values(ascending=False)
display(group_count)
```

```python
# Aggregate Passengers by Flight_ID
group_summary = df.groupby('Flight_ID')['Passengers'].agg(['count', 'mean', 'min', 'max']).sort_values('mean', ascending=False)
display(group_summary)
```

```python
# Cross-group record counts
cross_group = df.groupby(['Flight_ID', 'Flight_Date']).size().sort_values(ascending=False)
display(cross_group.head(15))
```

## 8. Data Transformations

Create useful derived information using Pandas transformations. Numerical columns are transformed into normalized and categorical representations where suitable.

```python
# Create a normalized version of Passengers
min_value = df['Passengers'].min()
max_value = df['Passengers'].max()

if max_value != min_value:
    df['Passengers_Normalized'] = (df['Passengers'] - min_value) / (max_value - min_value)
else:
    df['Passengers_Normalized'] = 0

df[['Passengers', 'Passengers_Normalized']].head()
```

```python
# Categorize Passengers into Low, Medium and High using quantiles
df['Passengers_Category'] = pd.qcut(
    df['Passengers'],
    q=3,
    labels=['Low', 'Medium', 'High'],
    duplicates='drop'
)

df['Passengers_Category'].value_counts().sort_index()
```

## 9. Meaningful Questions and Analysis

The following questions use Pandas operations to identify useful patterns in the flight operations data.

```python
### Question 1: Which Flight_ID category has the most records?
category_counts = df['Flight_ID'].value_counts()
display(category_counts.head(10))

print('Most common category:', category_counts.index[0])
print('Number of records:', category_counts.iloc[0])
```

```python
### Question 2: Which Flight_ID category has the highest average Passengers?
avg_by_category = df.groupby('Flight_ID')['Passengers'].mean().sort_values(ascending=False)
display(avg_by_category)

print('Highest average category:', avg_by_category.index[0])
print('Average Passengers:', round(avg_by_category.iloc[0], 2))
```

```python
### Question 3: What are the most frequent combinations of Flight_ID and Flight_Date?
combinations = df.groupby(['Flight_ID', 'Flight_Date']).size().sort_values(ascending=False)
display(combinations.head(10))
```

```python
### Question 4: What are the highest and lowest observed Passengers values?
print('Minimum Passengers:', df['Passengers'].min())
print('Maximum Passengers:', df['Passengers'].max())
print('Mean Passengers:', round(df['Passengers'].mean(), 2))
print('Median Passengers:', df['Passengers'].median())
```

## 10. Key Observations

The following observations summarize the main patterns identified from the analysis. The values are calculated directly from the dataset so the conclusions remain data-driven.

```python
# Generate 5–8 data-driven observations automatically from the actual dataset

observations = []

observations.append(
    f"1. The dataset contains {df.shape[0]} rows and {df.shape[1]} columns."
)

missing_total = int(df.isna().sum().sum())
if missing_total == 0:
    observations.append("2. No missing values were found in the dataset.")
else:
    observations.append(
        f"2. The dataset contains {missing_total} missing values across "
        f"{int((df.isna().sum() > 0).sum())} columns."
    )

duplicates = int(df.duplicated().sum())
observations.append(
    f"3. The dataset contains {duplicates} duplicate rows."
)

counts = df['Flight_ID'].value_counts()
observations.append(
    f"4. The most common Flight_ID category is '{counts.index[0]}' with "
    f"{counts.iloc[0]} records."
)

observations.append(
    f"5. The average Passengers is {df['Passengers'].mean():.2f}, while the median is "
    f"{df['Passengers'].median():.2f}."
)

observations.append(
    f"6. The minimum observed Passengers is {df['Passengers'].min():.2f} and the maximum "
    f"is {df['Passengers'].max():.2f}."
)

avg_groups = df.groupby('Flight_ID')['Passengers'].mean().sort_values(ascending=False)
observations.append(
    f"7. The Flight_ID category with the highest average Passengers is "
    f"'{avg_groups.index[0]}' with an average of {avg_groups.iloc[0]:.2f}."
)

combo = df.groupby(['Flight_ID', 'Flight_Date']).size().sort_values(ascending=False)
observations.append(
    f"8. The most frequent Flight_ID–Flight_Date combination is "
    f"'{combo.index[0][0]} – {combo.index[0][1]}' with {combo.iloc[0]} records."
)

for observation in observations:
    print(observation)
```
