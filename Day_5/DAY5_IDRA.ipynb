```python

# Day 5 - Student Grade Management System
# Using Object-Oriented Programming

import csv
import os

FILE_NAME = "students.csv"


class Student:
    """Represents a student and manages their academic details."""

    def __init__(self, student_id, name, age, course, marks):
        self.student_id = student_id
        self.name = name
        self.age = age
        self.course = course
        self.marks = marks

    # Calculate grade based on marks
    def calculate_grade(self):
        if self.marks >= 90:
            return "A+"
        elif self.marks >= 80:
            return "A"
        elif self.marks >= 70:
            return "B"
        elif self.marks >= 60:
            return "C"
        elif self.marks >= 50:
            return "D"
        else:
            return "F"

    # Return student details as a dictionary
    def to_dictionary(self):
        return {
            "ID": self.student_id,
            "Name": self.name,
            "Age": self.age,
            "Course": self.course,
            "Marks": self.marks,
            "Grade": self.calculate_grade()
        }

    # Display student details
    def display(self):
        print(f"ID     : {self.student_id}")
        print(f"Name   : {self.name}")
        print(f"Age    : {self.age}")
        print(f"Course : {self.course}")
        print(f"Marks  : {self.marks}")
        print(f"Grade  : {self.calculate_grade()}")
        print("-" * 35)


class StudentGradeManager:
    """Manages all student records."""

    def __init__(self):
        self.students = []
        self.courses = set()
        self.load_students()

    # Create CSV file if it doesn't exist
    def initialize_file(self):
        if not os.path.exists(FILE_NAME):
            with open(FILE_NAME, "w", newline="") as file:
                writer = csv.writer(file)
                writer.writerow(
                    ["ID", "Name", "Age", "Course", "Marks", "Grade"]
                )

    # Load students from CSV file
    def load_students(self):
        self.initialize_file()

        try:
            with open(FILE_NAME, "r", newline="") as file:
                reader = csv.DictReader(file)

                for row in reader:
                    student = Student(
                        row["ID"],
                        row["Name"],
                        int(row["Age"]),
                        row["Course"],
                        float(row["Marks"])
                    )

                    self.students.append(student)
                    self.courses.add(student.course)

        except (ValueError, KeyError):
            print("Warning: Some records in the file could not be loaded.")

    # Save all students to CSV
    def save_students(self):
        with open(FILE_NAME, "w", newline="") as file:
            fieldnames = ["ID", "Name", "Age", "Course", "Marks", "Grade"]
            writer = csv.DictWriter(file, fieldnames=fieldnames)

            writer.writeheader()

            for student in self.students:
                writer.writerow(student.to_dictionary())

    # Add a new student
    def add_student(self):
        print("\n--- Add Student ---")

        student_id = input("Enter Student ID: ").strip()

        # Check duplicate ID
        for student in self.students:
            if student.student_id == student_id:
                print("Student ID already exists!")
                return

        name = input("Enter Student Name: ").strip()

        if not name:
            print("Name cannot be empty.")
            return

        try:
            age = int(input("Enter Age: "))

            if age <= 0:
                print("Age must be greater than zero.")
                return

            course = input("Enter Course: ").strip()

            if not course:
                print("Course cannot be empty.")
                return

            marks = float(input("Enter Marks (0-100): "))

            if marks < 0 or marks > 100:
                print("Marks must be between 0 and 100.")
                return

            student = Student(
                student_id,
                name,
                age,
                course,
                marks
            )

            self.students.append(student)
            self.courses.add(course)

            self.save_students()

            print("Student added successfully!")
            print(f"Grade: {student.calculate_grade()}")

        except ValueError:
            print("Invalid input! Please enter valid numeric values.")

    # View all students
    def view_students(self):
        print("\n--- All Student Records ---")

        if not self.students:
            print("No student records found.")
            return

        for student in self.students:
            student.display()

    # Search student by ID or name
    def search_student(self):
        search_value = input(
            "Enter Student ID or Name: "
        ).strip().lower()

        found = False

        for student in self.students:
            if (student.student_id.lower() == search_value or
                    student.name.lower() == search_value):

                student.display()
                found = True

        if not found:
            print("Student not found.")

    # Update student details
    def update_student(self):
        student_id = input("Enter Student ID to update: ").strip()

        for student in self.students:
            if student.student_id == student_id:

                print("\nLeave a field empty to keep its current value.")

                name = input(
                    f"Name [{student.name}]: "
                ).strip()

                age = input(
                    f"Age [{student.age}]: "
                ).strip()

                course = input(
                    f"Course [{student.course}]: "
                ).strip()

                marks = input(
                    f"Marks [{student.marks}]: "
                ).strip()

                try:
                    if name:
                        student.name = name

                    if age:
                        new_age = int(age)

                        if new_age <= 0:
                            print("Invalid age.")
                            return

                        student.age = new_age

                    if course:
                        student.course = course
                        self.courses.add(course)

                    if marks:
                        new_marks = float(marks)

                        if new_marks < 0 or new_marks > 100:
                            print("Marks must be between 0 and 100.")
                            return

                        student.marks = new_marks

                    self.save_students()

                    print("Student details updated successfully!")
                    return

                except ValueError:
                    print("Invalid input. Update failed.")
                    return

        print("Student not found.")

    # Delete student
    def delete_student(self):
        student_id = input("Enter Student ID to delete: ").strip()

        for student in self.students:
            if student.student_id == student_id:

                self.students.remove(student)
                self.save_students()

                print("Student deleted successfully!")
                return

        print("Student not found.")

    # Display class statistics
    def show_statistics(self):
        if not self.students:
            print("No student records available.")
            return

        marks = [student.marks for student in self.students]

        total = sum(marks)
        average = total / len(marks)
        highest = max(marks)
        lowest = min(marks)

        topper = max(
            self.students,
            key=lambda student: student.marks
        )

        print("\n--- Class Statistics ---")
        print(f"Number of Students : {len(self.students)}")
        print(f"Average Marks      : {average:.2f}")
        print(f"Highest Marks      : {highest}")
        print(f"Lowest Marks       : {lowest}")

        print("\n--- Class Topper ---")
        print(f"Name   : {topper.name}")
        print(f"ID     : {topper.student_id}")
        print(f"Marks  : {topper.marks}")
        print(f"Grade  : {topper.calculate_grade()}")

    # Display available courses
    def show_courses(self):
        print("\n--- Available Courses ---")

        if not self.courses:
            print("No courses available.")
            return

        for course in sorted(self.courses):
            print(f"- {course}")

    # Main menu
    def menu(self):
        while True:
            print("\n" + "=" * 45)
            print("       STUDENT GRADE MANAGEMENT SYSTEM")
            print("=" * 45)

            print("1. Add Student")
            print("2. View All Students")
            print("3. Search Student")
            print("4. Update Student")
            print("5. Delete Student")
            print("6. Class Statistics")
            print("7. View Courses")
            print("8. Exit")

            choice = input("\nEnter your choice (1-8): ").strip()

            if choice == "1":
                self.add_student()

            elif choice == "2":
                self.view_students()

            elif choice == "3":
                self.search_student()

            elif choice == "4":
                self.update_student()

            elif choice == "5":
                self.delete_student()

            elif choice == "6":
                self.show_statistics()

            elif choice == "7":
                self.show_courses()

            elif choice == "8":
                print("\nThank you for using the Student Grade Management System!")
                break

            else:
                print("Invalid choice! Please enter a number from 1 to 8.")


if __name__ == "__main__":
    manager = StudentGradeManager()
    manager.menu()


```
