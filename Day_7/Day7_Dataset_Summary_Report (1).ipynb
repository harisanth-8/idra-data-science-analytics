# Dataset Summary Report – Student Dataset
 load and explore the Student Dataset. It demonstrates basic dataset inspection, structure, data types, descriptive statistics, and missing-value analysis.

```python
import pandas as pd

df = pd.read_csv('Day7_Student_Dataset.csv')
print('Dataset loaded successfully!')
print('Shape:', df.shape)
```

```python
rows, columns = df.shape
print('Number of rows:', rows)
print('Number of columns:', columns)
```

```python
print('Column names:')
for i, column in enumerate(df.columns, start=1):
    print(f'{i}. {column}')
```

```python
missing_values = df.isnull().sum()
missing_percentage = (df.isnull().sum() / len(df) * 100).round(2)
missing_summary = pd.DataFrame({'Missing Values': missing_values, 'Missing Percentage': missing_percentage})
missing_summary
```

```python
duplicate_count = df.duplicated().sum()
print('Number of duplicate rows:', duplicate_count)
```
